```python
import base64,hashlib,io,pathlib,subprocess,sys,zipfile
payload=base64.b64decode('')
assert hashlib.sha256(payload).hexdigest()=='51c6f2f6517c996a45f6fe64144468c40f2da9eb8129c23d928b2bb0f90017c1'
root=pathlib.Path("/kaggle/working/t3d-project")
root.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(payload)) as z:
    z.extractall(root)
```

```python
import torch
assert torch.cuda.is_available()
r=subprocess.run([sys.executable,"-m","unittest","discover","-s","tests","-v"],cwd=root,capture_output=True,text=True)
(root.parent/"tests.log").write_text(r.stdout+r.stderr)
print(r.stdout+r.stderr)
assert r.returncode==0
```
