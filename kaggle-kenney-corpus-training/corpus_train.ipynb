```python
import base64,hashlib,io,pathlib,subprocess,sys,zipfile,json
payload=base64.b64decode('')
assert hashlib.sha256(payload).hexdigest()=='e7b022825ad065ec16bbc96b75899965d205cfe312a18f2cdfdca49d38ffc7f8'
root=pathlib.Path("/kaggle/working/corpus-train");root.mkdir(exist_ok=True)
with zipfile.ZipFile(io.BytesIO(payload)) as z:z.extractall(root)
subprocess.run([sys.executable,"-m","pip","install","hilbertcurve>=2,<3"],check=True)
subprocess.run([sys.executable,"-m","tools.train_kenney_corpus_32"],cwd=root,check=True)
p=pathlib.Path("/kaggle/working/triton-kenney-corpus-32.json")
assert p.exists(),"Expected output json report"
r=json.loads(p.read_text())
r["source_sha256"]='e7b022825ad065ec16bbc96b75899965d205cfe312a18f2cdfdca49d38ffc7f8'
p.write_text(json.dumps(r,indent=2)+"\n")
```
